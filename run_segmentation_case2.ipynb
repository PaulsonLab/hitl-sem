# Segmentation, case study 2 — sequence of 16 tiles

Pixel-level segmentation of 16 tiles into four classes, annotated in sequence so
that each tile inherits what was learned from the previous ones.

Draw a handful of boxes per class on the top panel, press **Train & Predict** to
refit and refresh the class map and the entropy map, add boxes wherever the model
looks uncertain, and repeat. Press **Save & Next Image** to store the result and
move on; from the second image onward the boxes banked so far are used to
pre-segment the new image, so later images usually need only a correction or two.

Controls, with the cursor over the top panel: click & drag to draw a box,
`d` to delete the box under the cursor, `c` to cycle the active class.

`%matplotlib widget` is required — the dashboard needs an interactive canvas.

Pixel mode interpolates the patch features up to full resolution before
predicting, which is far heavier than patch mode: expect a few GB of RAM per
tile.


```python
%matplotlib widget

from hitl_sem.segmentation import SeqLoader
from hitl_sem.segmentation.dashboard import ActiveSegmentationDashboard
```

Directories. Features are extracted on first use and cached in `features_dir`.

```python
img_dir = './data/segmentation/case2/patches/'
features_dir = './data/segmentation/case2/features/'
labels_dir = './outputs/segmentation_case2/predictions/'
boxes_dir = './outputs/segmentation_case2/boxes/'
save_dir = './outputs/segmentation_case2/history/'
```

`mode='pixel'` predicts one label per pixel.

```python
loader = SeqLoader(img_dir, features_dir, labels_dir, boxes_dir)

dashboard = ActiveSegmentationDashboard(loader, save_dir, mode="pixel", track_progress=True)
```
